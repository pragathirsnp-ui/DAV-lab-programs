```python
data = {
    'Name': ['Janaki', 'Shikha', 'Sneha', 'Kiran', 'Meena', 'Kavya'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', None, 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df = pd.DataFrame(data)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)
```

```python
df['Department'] = df['Department'].ffill()
print(df)
```

```python
encoded_df = pd.get_dummies(df, columns=['Department'])
print(encoded_df)
```

```python
encoded_df = pd.get_dummies(df,columns=['Department'],dtype=int)
print(encoded_df)
```
