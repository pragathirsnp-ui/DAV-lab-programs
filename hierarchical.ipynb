Hierarchical Clustering using Mall Customers Dataset

step 1: import required libraries

```python
import matplotlib.pyplot as plt

import pandas as pd

from sklearn.preprocessing import StandardScaler

from sklearn.cluster import AgglomerativeClustering

from scipy.cluster.hierarchy import dendrogram, linkage
```

step 2: Create and load the dataset

```python
data = {
    'Annual Income': [15, 16, 17, 18, 19, 20, 55, 58, 60, 62,
                      65, 68, 85, 88, 90, 92, 95, 98, 30, 35],

    'Spending Score': [39, 42, 45, 40, 43, 48, 55, 58, 60, 52,
                       56, 59, 75, 78, 80, 72, 76, 82, 30, 35]
}

df = pd.DataFrame(data)

X = df[['Annual Income', 'Spending Score']]

print("Dataset shape:", X.shape)

print(X)
```

step 3: Standardize the data

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

step 4: Create a dendrogram

```python
linked = linkage(X_scaled, method='ward')

plt.figure(figsize=(12, 6))

dendrogram(linked)

plt.title("Dendrogram - Hierarchical Clustering")

plt.xlabel("Data Points")

plt.ylabel("Euclidean Distance")

plt.show()
```

step 5: Apply Agglomerative Clustering

```python
model = AgglomerativeClustering(n_clusters=3, linkage='ward')

clusters = model.fit_predict(X_scaled)
```

step 6: Visualize the clusters

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap='viridis',
    s=50
)

plt.xlabel("Standardized Annual Income")

plt.ylabel("Standardized Spending Score")

plt.title("Agglomerative Hierarchical Clustering")

plt.show()
```

step 7: Display cluster assignments

```python
print("\nCluster assignments:")

print(clusters)

print("\nFirst 10 data points and their clusters:")

for i in range(10):
    print("Data Point:", i, "Cluster:", clusters[i])
```
