Iris CSV file

Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

The Iris dataset is loaded from the Iris.csv file using the Pandas

```python
df = pd.read_csv(r"C:\Users\Admin\Downloads\archive (4)\Iris.csv")
```

SIZE od the dataset

```python
print(df.shape)
```

Distribution of Petal Length

A histogram is used to visualize the distribution and frequency of petal lengths in the Iris dataset

```python
plt.hist(df["PetalLengthCm"], bins=5)
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

Distribution of Petal Width

A histogram is used to visualize the distribution and frequency of petal widths in the Iris dataset.

```python
sns.histplot(data=df, x="PetalWidthCm", bins=15)

plt.title("Distribution of Petal width")
plt.xlabel("Petal Width (cm)")
plt.ylabel("Frequency")
plt.show()
```

Box Plot of Petal Length

A box plot is created for the petal length column to visualize its distribution and identify the median

```python
plt.boxplot(df["PetalLengthCm"])
plt.title("Iris Petal Length Box Plot (Matplotlib)")
plt.ylabel("Length (cm)")
plt.show()
```

Box Plot of Petal Width

A box plot is created for petal width to visualize its distribution and identify the median

```python
sns.boxplot(data=df, y="PetalWidthCm")
plt.title("Overall Petal Length Distribution")
plt.show()
```

Comparison of Petal Length Across Iris Species

A box plot is used to compare the petal length distributions of all three Iris species: Setosa, Versicolor, and Virginica

```python
sns.boxplot(data=df, x="Species", y="PetalLengthCm")
plt.title("Petal Length by Iris Species")
plt.show()
```

Petal Length vs Petal Width

A scatter plot is used to study the relationship between petal length and petal width

```python
plt.scatter(df["PetalLengthCm"], df["PetalWidthCm"]) 
plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

Petal Length vs Petal Width Using Seaborn

A Seaborn scatter plot is used to visualize the relationship between petal length and petal width for different Iris species

```python
sns.scatterplot(
    data=df,
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="Species"
)
plt.title("Petal Length vs Petal Width (Seaborn)")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

Pair Plot of Iris Features

A pair plot is used to visualize the relationships between the numerical features of the Iris dataset

```python
sns.pairplot(df.drop(columns=["Id"]), hue="Species", palette="Set1")
plt.show()
```

Correlation Heatmap

A correlation matrix is calculated for the numerical features of the Iris dataset and displayed using a heatmap

```python
numerical_df = df.select_dtypes("number").drop(columns=["Id"], errors="ignore")

correlation = numerical_df.corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
